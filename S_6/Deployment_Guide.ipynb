# Section 6: Streamlit Deployment Guide

### Task 1: Setup and `requirements.txt`

I have created the `app.py` script for trending movies using `pandas` and `requests`, alongside the `requirements.txt` file ensuring all packages are listed for cloud deployment.

### Task 2: Push to GitHub

To push this folder to a new public GitHub repository, execute these commands in your terminal from inside the `S_6` folder:

```bash
git init
git add app.py requirements.txt
git commit -m "Initial commit: Trending Movies App"
git branch -M main
git remote add origin https://github.com/YOUR_USERNAME/YOUR_REPO_NAME.git
git push -u origin main
```

### Task 3: Deploy via Streamlit Cloud

1. Go to [share.streamlit.io](https://share.streamlit.io/) and sign in with GitHub.
2. Click **New app** -> **Deploy a new app**.
3. Select the repository, branch (`main`), and main file path (`app.py`).
4. Click **Deploy!**

### Task 4: Continuous Deployment (CD)

To verify automatic redeployment:
1. Open `app.py` locally and change `st.title("Trending Movies App 🎬")` to `st.title("Trending Movies App (Live Update!) 🎬")`.
2. Run the following commands to push the update:
```bash
git add app.py
git commit -m "Updated title"
git push
```
3. Go back to your live Streamlit URL in the browser, refresh, and watch the title change without any manual redeployment.

### Task 5: Simulate a Deployment Error

To test error handling:
1. Open `requirements.txt` and delete the line containing `pandas`.
2. Commit and push the change:
```bash
git add requirements.txt
git commit -m "Removed pandas to simulate error"
git push
```
3. Watch the Streamlit Cloud dashboard. The build will fail, displaying a `ModuleNotFoundError: No module named 'pandas'`.
4. Add `pandas==2.2.0` back into `requirements.txt`.
5. Commit and push again to resolve the issue and restore the app to a working state.
